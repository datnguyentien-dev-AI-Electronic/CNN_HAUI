```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing import image


MODEL_PATH = "/content/CNN_NhandienHoa.h5"
IMAGE_PATH = "/content/tulip.jpg"
IMG_SIZE = (224, 224)  # Kích thước mô hình yêu cầu


LABELS = ["sunflower", "dandelion", "rose", "tulip"]

# 2. Tải mô hình
print("Đang tải mô hình...")
model = load_model(MODEL_PATH)
print("Tải mô hình thành công!")

# 3. Tiền xử lý ảnh đầu vào
# Tải ảnh và tự động resize về kích thước 224x224
img = image.load_img(IMAGE_PATH, target_size=IMG_SIZE)

# Chuyển ảnh thành mảng numpy (Kích thước: 224, 224, 3)
img_array = image.img_to_array(img)

# Thêm chiều batch_size (Kích thước sẽ thành: 1, 224, 224, 3)
img_array = np.expand_dims(img_array, axis=0)

# Chuẩn hóa giá trị pixel về khoảng [0, 1] giống như ImageDataGenerator(rescale=1./255) khi train
img_array = img_array / 255.0

# 4. Dự đoán
print("Đang tiến hành nhận diện...")
predictions = model.predict(img_array)

# 5. Xử lý kết quả đầu ra
predicted_class_idx = np.argmax(predictions, axis=1)[0]
confidence = predictions[0][predicted_class_idx] * 100

print("\n--- KẾT QUẢ NHẬN DIỆN ---")
print(f"Loại hoa dự đoán: {LABELS[predicted_class_idx]}")
print(f"Độ tin cậy (Confidence): {confidence:.2f}%")
print("Chi tiết xác suất các lớp:", predictions[0])
```
